#### **Outlier (1.5đ)**

```python
# kiểm tra giá trị outlier 
# Tạo df3 bao gồm các cột Q1, Q3, IQR, lower_bound, upper_bound, outlier_count (0.5 đ)
# ?
numeric_cols = df.select_dtypes(include="number").drop(columns=["id" ,"Churned"])
Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outlier_count = ((numeric_cols < lower_bound)|(numeric_cols > upper_bound)).sum()
df3 = pd.DataFrame({
    "Q1" : Q1 , 
    "Q3" : Q3 , 
    "IQR" : IQR,
    "lower_bound" :lower_bound , 
    "upper_bound" : upper_bound , 
    "outlier_count" : outlier_count 
})
df3
# df3 = ?
```

```python
# Xử lý outlier bằng 2 cách: IQR và clip
# df4 cho IQR (0.5 đ)

# df4 = df.copy()
# ?
df4 = df.copy()
mask = ((numeric_cols >= lower_bound)&(numeric_cols < upper_bound)).all(axis=1)
df4 = df4[mask]
# df5 cho clip (0.5 đ)
# df5 = df.copy()
# df5?
df5 = df.copy()
df5[numeric_cols.columns] = df5[numeric_cols.columns].clip(lower_bound, upper_bound, axis=1)
print(df4.shape)
print(df5.shape)
```

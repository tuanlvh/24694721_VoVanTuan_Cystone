### **Feature engineering**

Dummy variables for columns with few levels. Scale numeric columns (keep originals). Bin purchase amount.

```python
df = pd.get_dummies(df, columns=['Gender', 'Size', 'Category'], dtype=int)
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

minmax = MinMaxScaler()
standard = StandardScaler()

df['Amount_minmax'] = minmax.fit_transform(df[['Purchase Amount (USD)']])
df['Age_standard'] = standard.fit_transform(df[['Age']])
df[['Purchase Amount (USD)', 'Amount_minmax', 'Age', 'Age_standard']].head()
```

```python
df['Amount_bin'] = pd.cut(df['Purchase Amount (USD)'], bins=3, labels=['Low', 'Medium', 'High'])
df['Amount_bin'].value_counts()
```

# Insights
- Clothing is the most purchased category for both males and females, with items like blouses, jeans, and shirts among the top sellers.
- Males and females spend a broadly similar total amount overall, since males make up a larger share of the customer base.
- Average Review Rating for Clothing purchases is very close between males and females (around 3.7-3.8), showing no strong gender difference.
- Promo codes are used in roughly half of all transactions, suggesting price sensitivity is common across the customer base.
- Purchase Amount, Age, Review Rating, and Previous Purchases show weak correlation with each other, meaning spending amount isn't strongly predicted by these numeric features.

Caveat: correlation is not causation.
